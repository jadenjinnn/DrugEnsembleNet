```python
from pathlib import Path
import pandas as pd
from databases import NCBI, DrugBank

def get_genes_from_dataset_file(file_path):
        ncbi = NCBI()
        
        if (
            not file_path.exists()
            or not file_path.is_file()
            or file_path.suffix != ".csv"
        ):
            print(
                "%s either does not exist, is not a file, or is not a csv file",
                file_path,
            )

            return None

        genes = pd.read_csv(file_path).iloc[:, 0].tolist()

        genes_dict = {
            ncbi.check_symbol(gene): ncbi.get_id_by_symbol(gene)
            for gene in genes
            if ncbi.get_id_by_symbol(gene) and ncbi.check_symbol(gene)
        }

        return genes_dict
```

```python
def get_genes_from_directory_updown(directory):
    ncbi = NCBI()


    up_genes = set()
    down_genes = set()
    # print(gene_file_path)


    gene_file = pd.read_csv(directory)
    genes = pd.read_csv(directory).iloc[:, 0].tolist()
    log2FC = pd.read_csv(directory).iloc[:, 2].tolist()

    for gene, fc in zip(genes, log2FC):
        if fc > 0.25:
            up_genes.add(gene)
        elif fc < -0.25:
            down_genes.add(gene)


    up_genes_dict = {
        ncbi.check_symbol(gene): ncbi.get_id_by_symbol(gene)
        for gene in up_genes
        if ncbi.get_id_by_symbol(gene) and ncbi.check_symbol(gene)
    }

    down_genes_dict = {
        ncbi.check_symbol(gene): ncbi.get_id_by_symbol(gene)
        for gene in down_genes
        if ncbi.get_id_by_symbol(gene) and ncbi.check_symbol(gene)
    }

    return up_genes_dict, down_genes_dict

```
